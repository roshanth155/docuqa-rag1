```python
import re, time
import numpy as np
import pandas as pd
from pypdf import PdfReader
from rank_bm25 import BM25Okapi
from sentence_transformers import SentenceTransformer, CrossEncoder, util

PDF, QA = "syllabus.pdf", "qa.csv"
CHUNK_SIZE = 128   # words per chunk
THRESH = 0.65      # refusal cutoff, used in Step 4
SEED = 0
np.random.seed(SEED)

embedder = SentenceTransformer("BAAI/bge-small-en-v1.5")                 # dense search
reranker = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")          # reranker
```

```python
def tok(t): return re.findall(r"\w+", t.lower())
def norm(t): return re.sub(r"\s+", " ", t.lower()).strip()

def load_pages(pdf):
    return [(i + 1, p.extract_text() or "") for i, p in enumerate(PdfReader(pdf).pages)]

def chunk(pages, size, overlap=0.2):
    out, step = [], max(1, int(size * (1 - overlap)))
    for pg, text in pages:
        w = text.split()
        for s in range(0, max(len(w), 1), step):
            c = " ".join(w[s:s + size])
            if len(c.split()) > 15:
                out.append({"page": pg, "text": c})
    return out

class Index:
    def __init__(self, chunks):
        self.c = chunks
        self.texts = [c["text"] for c in chunks]
        self.bm = BM25Okapi([tok(t) for t in self.texts])
        self.emb = embedder.encode(self.texts, normalize_embeddings=True, convert_to_tensor=True)

    def bm25(self, q): return np.array(self.bm.get_scores(tok(q)))

    def dense(self, q):
        qe = embedder.encode("Represent this sentence for searching relevant passages: " + q,
                             normalize_embeddings=True, convert_to_tensor=True)
        return util.cos_sim(qe, self.emb)[0].cpu().numpy()

    def hybrid(self, q, a=0.5):
        z = lambda x: (x - x.mean()) / (x.std() + 1e-9)
        return a * z(self.dense(q)) + (1 - a) * z(self.bm25(q))

    def search(self, q, method, k):
        if method == "rerank":
            cand = np.argsort(-self.hybrid(q))[:20]
            sc = reranker.predict([(q, self.texts[i]) for i in cand])
            order = np.argsort(-sc)[:k]
            return [int(cand[i]) for i in order], [float(sc[i]) for i in order]
        sc = getattr(self, method)(q)
        idx = np.argsort(-sc)[:k]
        return [int(i) for i in idx], [float(sc[i]) for i in idx]
```

```python
pages = load_pages(PDF)
qa = pd.read_csv(QA)
qa_ans = qa[qa["answerable"] == 1].reset_index(drop=True)
print(f"{len(pages)} pages | {len(qa_ans)} answerable | {(qa.answerable == 0).sum()} out-of-scope")

BEST = Index(chunk(pages, CHUNK_SIZE))
print("chunks:", len(BEST.texts))

q = qa_ans.question[0]
idx, sc = BEST.search(q, "rerank", 3)
print("\nQ:", q)
for i in idx:
    print(f"\n[Page {BEST.c[i]['page']}] {BEST.texts[i][:300]}...")
```

```python
def evaluate(index, method, ks=(1, 3, 5, 10)):
    ranks, t0 = [], time.time()
    for _, r in qa_ans.iterrows():
        idx, _ = index.search(r["question"], method, max(ks))
        rank = next((i + 1 for i, j in enumerate(idx)
                     if norm(str(r["snippet"])) in norm(index.texts[j])), None)
        ranks.append(rank)
    lat = (time.time() - t0) / len(qa_ans) * 1000
    res = {f"Recall@{k}": np.mean([r is not None and r <= k for r in ranks]) for k in ks}
    res["MRR"] = np.mean([1 / r if r else 0 for r in ranks])
    res["Latency(ms/query)"] = lat
    return res, ranks
```

```python
methods = ["bm25", "dense", "hybrid", "rerank"]
rows, all_ranks = [], {}

for size in [128, 256, 512]:
    idx = Index(chunk(pages, size))
    for m in methods:
        res, ranks = evaluate(idx, m)
        all_ranks[(size, m)] = ranks
        rows.append({"chunk_size": size, "method": m, **res})

results = pd.DataFrame(rows).round(3)
print(results.to_string(index=False))
results.to_csv("results.csv", index=False)
```

```python
def bootstrap_mrr(ranks, n=2000):
    rr = np.array([1 / r if r else 0 for r in ranks])
    rng = np.random.default_rng(SEED)
    means = [rng.choice(rr, len(rr)).mean() for _ in range(n)]
    return rr.mean(), np.percentile(means, 2.5), np.percentile(means, 97.5)

ci_rows = []
for (size, m), ranks in all_ranks.items():
    mean, lo, hi = bootstrap_mrr(ranks)
    ci_rows.append({"chunk_size": size, "method": m, "MRR": round(mean, 3),
                    "CI_low": round(lo, 3), "CI_high": round(hi, 3)})
ci = pd.DataFrame(ci_rows)
print(ci.to_string(index=False))
ci.to_csv("mrr_ci.csv", index=False)
```

```python
from sklearn.metrics import roc_auc_score

BEST = Index(chunk(pages, CHUNK_SIZE))

def top_dense_score(q, index=None):
    return float(np.max((index or BEST).dense(q)))

ans_scores = [top_dense_score(q) for q in qa[qa.answerable == 1].question]
oos_scores = [top_dense_score(q) for q in qa[qa.answerable == 0].question]

print("Avg top score, answerable:  ", np.mean(ans_scores).round(3))
print("Avg top score, out-of-scope:", np.mean(oos_scores).round(3))

y = [1] * len(ans_scores) + [0] * len(oos_scores)
print("AUROC:", round(roc_auc_score(y, ans_scores + oos_scores), 3))
```

```python
rows = []
for t in [0.55, 0.60, 0.65, 0.68, 0.70, 0.72, 0.75]:
    rows.append({
        "THRESH": t,
        "Correctly refused OOS": round(np.mean([s < t for s in oos_scores]), 3),
        "Wrongly refused answerable": round(np.mean([s < t for s in ans_scores]), 3),
    })
thresh_table = pd.DataFrame(rows)
print(thresh_table.to_string(index=False))
thresh_table.to_csv("threshold_table.csv", index=False)
```

```python
import html, torch
import gradio as gr
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

dev = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", dev)   # should say cuda

t5_tok = AutoTokenizer.from_pretrained("google/flan-t5-large")
t5 = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-large").to(dev)

def gen(prompt):
    inputs = t5_tok(prompt, return_tensors="pt", truncation=True, max_length=1024).to(dev)
    out = t5.generate(**inputs, max_new_tokens=100, num_beams=3)
    return t5_tok.decode(out[0], skip_special_tokens=True)
```

```python
esc = html.escape
IDLE = '<div class="idle">Ask a question about the document.</div>'

def answer(question):
    if not question or not question.strip():
        return IDLE, ""
    t0 = time.time()
    conf = top_dense_score(question)

    if conf < THRESH:
        card = (f'<div class="card refuse"><div class="label">Not found</div>'
                f'<div class="ans">This question doesn\'t seem to be covered in the document.</div>'
                f'<div class="meta"><span class="chip">similarity {conf:.2f} &lt; threshold {THRESH}</span></div></div>')
        return card, ""

    idx, sc = BEST.search(question, "rerank", 3)
    ctx = "\n\n".join(BEST.texts[i] for i in idx)
    prompt = ("Read the rules below and answer the question in one complete sentence. "
              f"Use only the rules.\n\nRules:\n{ctx}\n\nQuestion: {question}\nAnswer:")
    out = gen(prompt)
    secs = time.time() - t0

    card = (f'<div class="card"><div class="label">Answer</div><div class="ans">{esc(out)}</div>'
            f'<div class="meta"><span class="chip">confidence {conf:.2f}</span>'
            f'<span class="chip">{len(idx)} sources</span>'
            f'<span class="chip">{secs:.1f} s</span></div></div>')
    srcs = ""
    for rank, (i, s) in enumerate(zip(idx, sc), 1):
        srcs += (f'<div class="src"><div class="top"><span class="badge">Source {rank} · Page {BEST.c[i]["page"]}</span>'
                 f'<span class="sc">rerank score {s:.2f}</span></div>'
                 f'<div class="txt">{esc(BEST.texts[i])}</div></div>')
    return card, srcs

def rebuild(file):
    global BEST
    if file is None:
        return "Waiting for a PDF..."
    path = file if isinstance(file, str) else file.name
    BEST = Index(chunk(load_pages(path), CHUNK_SIZE))
    return "✅ Document indexed. Ask away!"
```

```python
# ============================================================
# DocuQA: advanced demo UI (replaces Step 5, Cell 3)
# Needs from earlier cells: gen, BEST, top_dense_score, reranker, THRESH,
# CHUNK_SIZE, tok, Index, chunk, load_pages, pages, dev, and (optional)
# the `results` and `thresh_table` DataFrames from Steps 3 and 4.
# ============================================================
import html, re, math, time
import numpy as np
import pandas as pd
import gradio as gr

esc = html.escape
STOP = set("a an the of to in on for and or is are what how when which who does do be by with at from as it its "
           "this that if my i amp quot lt gt 39".split())

def sigmoid(x): return 1 / (1 + math.exp(-x))

# ---------------------------------------------------------------- CSS
CSS = """
@import url('https://fonts.googleapis.com/css2?family=Inter:wght@400;500;600;700;800&family=JetBrains+Mono:wght@400;600&display=swap');
:root{--bg:#06060d;--glass:rgba(255,255,255,.045);--line:rgba(255,255,255,.09);--txt:#e8e8f6;--mut:#9a9ab8;
--v:#8b5cf6;--c:#22d3ee;--p:#f472b6;--g:#34d399;--a:#fbbf24;--r:#fb7185}
body{background:var(--bg) !important}
body::before,body::after{content:"";position:fixed;z-index:-1;width:60vmax;height:60vmax;border-radius:50%;filter:blur(90px);opacity:.38;pointer-events:none}
body::before{top:-25vmax;left:-15vmax;background:radial-gradient(circle,#6d28d9,transparent 65%);animation:drift1 22s ease-in-out infinite alternate}
body::after{bottom:-30vmax;right:-20vmax;background:radial-gradient(circle,#0891b2,transparent 65%);animation:drift2 26s ease-in-out infinite alternate}
@keyframes drift1{to{transform:translate(18vmax,14vmax) scale(1.2)}}
@keyframes drift2{to{transform:translate(-16vmax,-12vmax) scale(1.15)}}
.gradio-container{max-width:1180px !important;margin:auto;background:transparent !important;font-family:'Inter',sans-serif !important;color:var(--txt)}
.gradio-container *{--body-text-color:var(--txt);--block-label-text-color:var(--mut)}
.gradio-container .block,.gradio-container .form,.gradio-container .tabitem{background:transparent !important;border:none !important;box-shadow:none !important}
.gradio-container textarea,.gradio-container input[type=text]{background:rgba(255,255,255,.05) !important;color:var(--txt) !important;border:1px solid var(--line) !important;border-radius:14px !important;font-size:16px !important;padding:14px 16px !important}
.gradio-container textarea:focus,.gradio-container input:focus{border-color:var(--v) !important;box-shadow:0 0 0 3px rgba(139,92,246,.25) !important}
.gradio-container button.primary{background:linear-gradient(135deg,#7c3aed,#06b6d4) !important;border:none !important;border-radius:14px !important;font-weight:700 !important;box-shadow:0 8px 28px rgba(124,58,237,.4);transition:transform .15s}
.gradio-container button.primary:hover{transform:translateY(-2px)}
.gradio-container button.secondary{background:var(--glass) !important;border:1px solid var(--line) !important;color:var(--txt) !important;border-radius:14px !important}
.gradio-container .tab-nav button,.gradio-container [role=tab]{color:var(--mut) !important;font-weight:600 !important}
.gradio-container [role=tab][aria-selected=true]{color:#fff !important;border-color:var(--v) !important}
.gradio-container table.examples,.gradio-container .examples button{background:var(--glass) !important;color:var(--txt) !important;border-color:var(--line) !important}

/* hero */
.hero{position:relative;padding:34px 38px;border-radius:26px;overflow:hidden;background:linear-gradient(135deg,rgba(124,58,237,.35),rgba(6,182,212,.18));border:1px solid var(--line);backdrop-filter:blur(16px)}
.hero::after{content:"";position:absolute;inset:0;background:repeating-linear-gradient(115deg,transparent 0 22px,rgba(255,255,255,.025) 22px 23px);pointer-events:none}
.hero h1{margin:0;font-size:38px;font-weight:800;letter-spacing:-1px;background:linear-gradient(90deg,#fff,#c4b5fd 45%,#67e8f9);-webkit-background-clip:text;background-clip:text;color:transparent}
.hero p{margin:8px 0 0;color:#c9c9e6;font-size:15.5px;max-width:720px}
.badges{display:flex;flex-wrap:wrap;gap:8px;margin-top:18px}
.badge-p{font-size:12.5px;font-weight:600;padding:6px 13px;border-radius:999px;background:rgba(255,255,255,.09);border:1px solid rgba(255,255,255,.18)}
.live{display:inline-block;width:8px;height:8px;border-radius:50%;background:var(--g);margin-right:7px;box-shadow:0 0 0 0 rgba(52,211,153,.7);animation:ping 1.8s infinite}
@keyframes ping{70%{box-shadow:0 0 0 10px rgba(52,211,153,0)}100%{box-shadow:0 0 0 0 rgba(52,211,153,0)}}

/* stats */
.stats{display:grid;grid-template-columns:repeat(4,1fr);gap:12px;margin:14px 0 4px}
.stat{padding:16px 18px;border-radius:18px;background:var(--glass);border:1px solid var(--line);backdrop-filter:blur(12px)}
.stat .n{font-size:26px;font-weight:800;font-family:'JetBrains Mono',monospace;background:linear-gradient(90deg,#c4b5fd,#67e8f9);-webkit-background-clip:text;background-clip:text;color:transparent}
.stat .l{font-size:11.5px;letter-spacing:1.1px;text-transform:uppercase;color:var(--mut);margin-top:2px}
@media(max-width:760px){.stats{grid-template-columns:repeat(2,1fr)}}

/* pipeline */
.pipe{display:flex;align-items:flex-start;justify-content:space-between;padding:20px 16px;border-radius:20px;background:var(--glass);border:1px solid var(--line);backdrop-filter:blur(12px);margin-top:10px}
.step{flex:0 0 auto;width:118px;text-align:center;transition:all .3s}
.orb{width:50px;height:50px;margin:0 auto 8px;border-radius:50%;display:grid;place-items:center;font-size:21px;background:rgba(255,255,255,.06);border:1.5px solid var(--line)}
.sname{font-weight:700;font-size:13.5px}.sdesc{font-size:10.5px;color:var(--mut);margin-top:2px;min-height:26px}
.stime{font-family:'JetBrains Mono',monospace;font-size:11.5px;margin-top:4px;color:var(--mut)}
.step.pending{opacity:.45}
.step.active .orb{border-color:var(--v);background:rgba(139,92,246,.25);animation:pulse 1.1s infinite}
.step.done .orb{border-color:var(--g);background:rgba(52,211,153,.18);box-shadow:0 0 18px rgba(52,211,153,.35)}
.step.done .stime{color:var(--g)}
.step.stop .orb{border-color:var(--a);background:rgba(251,191,36,.18);box-shadow:0 0 18px rgba(251,191,36,.35)}
.step.stop .stime{color:var(--a)}
.step.skip{opacity:.25}
@keyframes pulse{0%,100%{box-shadow:0 0 0 0 rgba(139,92,246,.6)}50%{box-shadow:0 0 0 12px rgba(139,92,246,0)}}
.link{flex:1;height:2px;margin-top:25px;background:var(--line);border-radius:2px;position:relative;overflow:hidden}
.link.done{background:linear-gradient(90deg,var(--g),var(--c))}
.link.active::after{content:"";position:absolute;inset:0;background:linear-gradient(90deg,transparent,var(--v),transparent);animation:flow 1s linear infinite}
@keyframes flow{from{transform:translateX(-100%)}to{transform:translateX(100%)}}
@media(max-width:760px){.step{width:70px}.sdesc{display:none}}

/* answer card */
.card{border-radius:22px;padding:22px 24px;margin-top:12px;background:linear-gradient(135deg,rgba(124,58,237,.16),rgba(6,182,212,.08));border:1px solid rgba(139,92,246,.35);backdrop-filter:blur(14px);animation:rise .45s ease}
.card.refuse{background:linear-gradient(135deg,rgba(251,191,36,.14),rgba(251,113,133,.07));border-color:rgba(251,191,36,.4)}
@keyframes rise{from{opacity:0;transform:translateY(12px)}to{opacity:1;transform:none}}
.card .row{display:flex;gap:18px;align-items:center}.grow{flex:1}
.label{font-size:11px;font-weight:700;letter-spacing:1.6px;text-transform:uppercase;color:#a78bfa;margin-bottom:8px}
.refuse .label{color:var(--a)}
.ans{font-size:21px;line-height:1.5;font-weight:650}
.meta{display:flex;gap:8px;flex-wrap:wrap;margin-top:14px}
.chip{font-size:12px;font-weight:600;padding:5px 11px;border-radius:9px;background:rgba(34,211,238,.1);border:1px solid rgba(34,211,238,.3);font-family:'JetBrains Mono',monospace}
.ring{width:96px;height:96px;flex:0 0 auto}
.rbg{fill:none;stroke:rgba(255,255,255,.1);stroke-width:7}
.rfg{fill:none;stroke:url(#rg);stroke-width:7;stroke-linecap:round;transition:stroke-dasharray .8s}
.rtxt{fill:#fff;font-size:17px;font-weight:800;font-family:'JetBrains Mono',monospace}
.rcap{font-size:10px;text-align:center;color:var(--mut);letter-spacing:1px;text-transform:uppercase;margin-top:-4px}
.think{display:flex;align-items:center;gap:12px;padding:20px 24px;margin-top:12px;border-radius:22px;background:var(--glass);border:1px dashed var(--line);color:var(--mut)}
.dots span{display:inline-block;width:8px;height:8px;margin:0 2px;border-radius:50%;background:var(--v);animation:bounce 1s infinite}
.dots span:nth-child(2){animation-delay:.15s}.dots span:nth-child(3){animation-delay:.3s}
@keyframes bounce{0%,80%,100%{transform:scale(.6);opacity:.5}40%{transform:scale(1.1);opacity:1}}
.idle{padding:26px;text-align:center;color:var(--mut);border:1px dashed var(--line);border-radius:22px;margin-top:12px}

/* sources */
.src{border-radius:18px;padding:16px 18px;margin-top:10px;background:var(--glass);border:1px solid var(--line);backdrop-filter:blur(10px);animation:rise .5s ease}
.src .top{display:flex;justify-content:space-between;align-items:center;gap:10px;margin-bottom:10px}
.badge{background:linear-gradient(135deg,#7c3aed,#06b6d4);color:#fff;font-size:12px;font-weight:700;padding:4px 12px;border-radius:999px}
.bar{height:6px;border-radius:6px;background:rgba(255,255,255,.08);flex:1;max-width:160px;overflow:hidden}
.bar i{display:block;height:100%;border-radius:6px;background:linear-gradient(90deg,var(--v),var(--c))}
.sc{font-size:11.5px;color:var(--mut);font-family:'JetBrains Mono',monospace;white-space:nowrap}
.src .txt{font-size:13.8px;line-height:1.65;color:#cfcfe8;max-height:150px;overflow-y:auto}
mark{background:rgba(251,191,36,.25);color:#fde68a;padding:0 3px;border-radius:4px}
.sec{font-size:12px;font-weight:700;letter-spacing:1.5px;text-transform:uppercase;color:var(--mut);margin:20px 4px 2px}

/* history */
.hist{display:flex;flex-direction:column;gap:8px}
.hi{padding:11px 14px;border-radius:14px;background:var(--glass);border:1px solid var(--line);font-size:13px}
.hi .q{font-weight:600;margin-bottom:5px}.hi .pill{font-size:10.5px;font-weight:700;padding:2px 8px;border-radius:999px;margin-right:6px}
.pill.ok{background:rgba(52,211,153,.18);color:var(--g)}.pill.no{background:rgba(251,191,36,.18);color:var(--a)}
.hi .m{font-family:'JetBrains Mono',monospace;font-size:11px;color:var(--mut)}
.empty{color:var(--mut);font-size:13px;padding:10px 4px}

/* benchmarks */
.bgrid{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}
@media(max-width:860px){.bgrid{grid-template-columns:1fr}}
.bcard{padding:18px;border-radius:20px;background:var(--glass);border:1px solid var(--line)}
.bcard h4{margin:0 0 12px;font-size:14px}
.brow{display:grid;grid-template-columns:62px 1fr 48px;gap:8px;align-items:center;margin:8px 0;font-size:12.5px}
.brow .t{height:10px;border-radius:8px;background:rgba(255,255,255,.07);overflow:hidden}
.brow .t i{display:block;height:100%;border-radius:8px}
.brow b{font-family:'JetBrains Mono',monospace;font-size:12px;text-align:right}
.best{color:var(--g)}
.lat{font-size:10.5px;color:var(--mut);margin-left:62px;margin-top:-4px}
.how{display:grid;grid-template-columns:repeat(2,1fr);gap:12px}
@media(max-width:760px){.how{grid-template-columns:1fr}}
.hcard{padding:20px;border-radius:20px;background:var(--glass);border:1px solid var(--line)}
.hcard h4{margin:0 0 6px;font-size:15px}.hcard p{margin:0;color:#c1c1dc;font-size:13.5px;line-height:1.6}
.hcard .num{font-family:'JetBrains Mono',monospace;color:var(--c);font-size:12px;margin-bottom:6px}
svg text{font-family:'Inter',sans-serif}
"""

# ------------------------------------------------------------ HTML builders
HERO = """
<div class="hero">
  <h1>📚 DocuQA</h1>
  <p>Ask a question about a document and get a grounded one-sentence answer with page-level citations.
  Questions the document can't answer are refused instead of hallucinated.</p>
  <div class="badges">
    <span class="badge-p"><span class="live"></span>Live</span>
    <span class="badge-p">🔎 Hybrid BM25 + Dense</span>
    <span class="badge-p">🎯 Cross-encoder rerank</span>
    <span class="badge-p">🧠 Flan-T5-large</span>
    <span class="badge-p">🛡️ Refusal guard</span>
  </div>
</div>"""

IDLE = '<div class="idle">✨ Ask a question to watch the pipeline run, step by step.</div>'
THINKING = ('<div class="think"><div class="dots"><span></span><span></span><span></span></div>'
            'Working on it…</div>')

def _best_mrr():
    try:
        r = get_results()
        return f"{r['MRR'].max():.3f}"
    except Exception:
        return "–"

def stats_html():
    return (f'<div class="stats">'
            f'<div class="stat"><div class="n">{len(pages)}</div><div class="l">Pages indexed</div></div>'
            f'<div class="stat"><div class="n">{len(BEST.texts)}</div><div class="l">Chunks · {CHUNK_SIZE} words</div></div>'
            f'<div class="stat"><div class="n">{_best_mrr()}</div><div class="l">Best retrieval MRR</div></div>'
            f'<div class="stat"><div class="n">{str(dev).upper()}</div><div class="l">Compute · τ = {THRESH}</div></div>'
            f'</div>')

STAGES = [("Guard", "🛡️", "Similarity vs. threshold"),
          ("Retrieve", "🔎", "BM25 + dense, top 20"),
          ("Rerank", "🎯", "Cross-encoder → top 3"),
          ("Generate", "🧠", "Flan-T5 answer")]

def pipeline_html(status, ms):
    out = '<div class="pipe">'
    for i, ((name, icon, desc), st) in enumerate(zip(STAGES, status)):
        t = f"{ms[i]:.0f} ms" if ms[i] is not None else "·"
        out += (f'<div class="step {st}"><div class="orb">{icon}</div><div class="sname">{name}</div>'
                f'<div class="sdesc">{desc}</div><div class="stime">{t}</div></div>')
        if i < len(STAGES) - 1:
            nxt = "active" if st == "done" and status[i + 1] == "active" else ("done" if st == "done" and status[i + 1] in ("done",) else "")
            out += f'<div class="link {nxt}"></div>'
    return out + "</div>"

def ring_html(conf):
    pct = max(0.0, min(1.0, (conf - 0.5) / 0.4))
    C = 2 * math.pi * 34
    return ('<div><svg class="ring" viewBox="0 0 80 80"><defs><linearGradient id="rg" x1="0" y1="0" x2="1" y2="1">'
            '<stop offset="0" stop-color="#8b5cf6"/><stop offset="1" stop-color="#22d3ee"/></linearGradient></defs>'
            '<circle cx="40" cy="40" r="34" class="rbg"/>'
            f'<circle cx="40" cy="40" r="34" class="rfg" stroke-dasharray="{C*pct:.1f} {C:.1f}" transform="rotate(-90 40 40)"/>'
            f'<text x="40" y="46" text-anchor="middle" class="rtxt">{conf:.2f}</text></svg>'
            '<div class="rcap">match</div></div>')

def answer_card(out, conf, n_src, secs, note=""):
    return (f'<div class="card"><div class="row"><div class="grow"><div class="label">Answer</div>'
            f'<div class="ans">{esc(out)}</div><div class="meta">'
            f'<span class="chip">sim {conf:.3f}</span><span class="chip">{n_src} sources</span>'
            f'<span class="chip">{secs:.1f}s total</span>{note}</div></div>{ring_html(conf)}</div></div>')

def refuse_card(conf, tau, rr):
    return (f'<div class="card refuse"><div class="row"><div class="grow"><div class="label">Not found in document</div>'
            f'<div class="ans">I couldn\'t find a confident answer to that in this document.</div>'
            f'<div class="meta"><span class="chip">sim {conf:.3f} &lt; τ {tau:.2f}</span>'
            f'<span class="chip">best rerank {rr:+.2f}</span>'
            f'<span class="chip">tip: lower τ or switch the guard off</span></div></div>{ring_html(conf)}</div></div>')

def highlight(text, question):
    terms = {w for w in tok(question) if w not in STOP and len(w) > 2}
    t = esc(text)
    if not terms:
        return t
    pat = re.compile(r"\b(" + "|".join(sorted(map(re.escape, terms), key=len, reverse=True)) + r")\b", re.I)
    return pat.sub(r"<mark>\1</mark>", t)

def sources_html(idx, scores, question, title="Cited sources"):
    out = f'<div class="sec">{title}</div>'
    for rank, (i, s) in enumerate(zip(idx, scores), 1):
        rel = sigmoid(s) * 100
        out += (f'<div class="src"><div class="top"><span class="badge">Source {rank} · Page {BEST.c[i]["page"]}</span>'
                f'<div class="bar"><i style="width:{rel:.0f}%"></i></div>'
                f'<span class="sc">rerank {s:+.2f}</span></div>'
                f'<div class="txt">{highlight(BEST.texts[i], question)}</div></div>')
    return out

def history_html(hist):
    if not hist:
        return '<div class="empty">Your questions will appear here.</div>'
    out = '<div class="hist">'
    for h in reversed(hist[-7:]):
        pill = '<span class="pill ok">answered</span>' if h["ok"] else '<span class="pill no">refused</span>'
        out += (f'<div class="hi"><div class="q">{esc(h["q"])}</div>{pill}'
                f'<span class="m">sim {h["conf"]:.2f} · {h["secs"]:.1f}s</span></div>')
    return out + "</div>"

def get_results():
    try:
        return results.copy()
    except NameError:
        return pd.read_csv("results.csv")

def get_thresh():
    try:
        return thresh_table.copy()
    except NameError:
        return pd.read_csv("threshold_table.csv")

def benchmarks_html():
    r = get_results()
    best = r.loc[r["MRR"].idxmax()]
    cols = {"bm25": "#fbbf24", "dense": "#22d3ee", "hybrid": "#a78bfa", "rerank": "#34d399"}
    out = '<div class="sec">Retrieval benchmark · MRR by chunk size and method</div><div class="bgrid">'
    for size in sorted(r["chunk_size"].unique()):
        out += f'<div class="bcard"><h4>{int(size)}-word chunks</h4>'
        for _, row in r[r["chunk_size"] == size].iterrows():
            is_best = (row["chunk_size"] == best["chunk_size"] and row["method"] == best["method"])
            cls = "best" if is_best else ""
            crown = " 👑" if is_best else ""
            out += (f'<div class="brow"><span class="{cls}">{row["method"]}{crown}</span>'
                    f'<div class="t"><i style="width:{row["MRR"]*100:.0f}%;background:{cols.get(row["method"], "#999")}"></i></div>'
                    f'<b class="{cls}">{row["MRR"]:.3f}</b></div>'
                    f'<div class="lat">R@1 {row["Recall@1"]:.2f} · R@5 {row["Recall@5"]:.2f} · {row["Latency(ms/query)"]:.0f} ms/query</div>')
        out += "</div>"
    return out + "</div>"

def threshold_svg():
    t = get_thresh().sort_values("THRESH")
    W, H, L, B, T = 640, 260, 54, 40, 18
    xs = t["THRESH"].values
    def X(v): return L + (v - xs.min()) / (xs.max() - xs.min()) * (W - L - 20)
    def Y(v): return T + (1 - v) * (H - T - B)
    def line(col, color):
        pts = " ".join(f"{X(a):.1f},{Y(b):.1f}" for a, b in zip(xs, t[col].values))
        dots = "".join(f'<circle cx="{X(a):.1f}" cy="{Y(b):.1f}" r="4" fill="{color}"/>' for a, b in zip(xs, t[col].values))
        return f'<polyline points="{pts}" fill="none" stroke="{color}" stroke-width="2.5"/>{dots}'
    grid = "".join(f'<line x1="{L}" x2="{W-20}" y1="{Y(v):.1f}" y2="{Y(v):.1f}" stroke="rgba(255,255,255,.08)"/>'
                   f'<text x="{L-8}" y="{Y(v)+4:.1f}" text-anchor="end" fill="#9a9ab8" font-size="11">{v:.1f}</text>'
                   for v in (0, .25, .5, .75, 1))
    ticks = "".join(f'<text x="{X(a):.1f}" y="{H-18}" text-anchor="middle" fill="#9a9ab8" font-size="11">{a:g}</text>' for a in xs)
    cur = ""
    if xs.min() <= THRESH <= xs.max():
        cur = (f'<line x1="{X(THRESH):.1f}" x2="{X(THRESH):.1f}" y1="{T}" y2="{H-B}" stroke="#f472b6" stroke-dasharray="5 4"/>'
               f'<text x="{X(THRESH)+6:.1f}" y="{T+12}" fill="#f472b6" font-size="11" font-weight="700">current τ = {THRESH}</text>')
    return ('<div class="sec">Refusal guard · threshold trade-off</div><div class="bcard">'
            f'<svg viewBox="0 0 {W} {H}" width="100%">{grid}{ticks}{cur}'
            f'{line("Correctly refused OOS", "#34d399")}{line("Wrongly refused answerable", "#fb7185")}'
            f'<text x="{L}" y="{H-2}" fill="#9a9ab8" font-size="11">threshold τ (dense cosine)</text>'
            f'<g font-size="12"><circle cx="{W-250}" cy="{T+4}" r="5" fill="#34d399"/><text x="{W-240}" y="{T+8}" fill="#e8e8f6">out-of-scope refused</text>'
            f'<circle cx="{W-250}" cy="{T+24}" r="5" fill="#fb7185"/><text x="{W-240}" y="{T+28}" fill="#e8e8f6">answerable wrongly refused</text></g>'
            '</svg></div>')

HOW = """
<div class="sec">How DocuQA works</div>
<div class="how">
 <div class="hcard"><div class="num">01 · GUARD</div><h4>Is the question in scope?</h4>
  <p>The question is embedded and compared to every chunk. If the best cosine similarity is below τ, the system says "not found" and never calls the generator.</p></div>
 <div class="hcard"><div class="num">02 · RETRIEVE</div><h4>Hybrid search</h4>
  <p>BM25 catches exact terms like "I grade" or "85%". Dense embeddings catch paraphrases. Their z-scored scores are mixed 50/50 to get the top 20 chunks.</p></div>
 <div class="hcard"><div class="num">03 · RERANK</div><h4>Cross-encoder precision</h4>
  <p>A cross-encoder reads the question and each candidate together, which is slower but far more precise, and keeps the 3 best chunks.</p></div>
 <div class="hcard"><div class="num">04 · GENERATE</div><h4>Grounded answer</h4>
  <p>Flan-T5-large writes one sentence using only those 3 chunks. Sources are shown with the query terms highlighted so every answer can be verified.</p></div>
</div>"""

# ------------------------------------------------------------ app logic
RR_RESCUE = 0.0   # cross-encoder logit above which a low-similarity question is still answered

def ask(question, hist, tau=0.60, guard_off=False):
    hist = list(hist or [])
    ms = [None] * 4
    if not question or not question.strip():
        yield IDLE, pipeline_html(["pending"] * 4, ms), "", history_html(hist), hist
        return
    t_all = time.time()

    yield THINKING, pipeline_html(["active", "pending", "pending", "pending"], ms), "", history_html(hist), hist
    t = time.time(); conf = top_dense_score(question); ms[0] = (time.time() - t) * 1000

    yield THINKING, pipeline_html(["done", "active", "pending", "pending"], ms), "", history_html(hist), hist
    t = time.time(); cand = np.argsort(-BEST.hybrid(question))[:20]; ms[1] = (time.time() - t) * 1000

    yield THINKING, pipeline_html(["done", "done", "active", "pending"], ms), "", history_html(hist), hist
    t = time.time()
    sc = reranker.predict([(question, BEST.texts[i]) for i in cand])
    order = np.argsort(-sc)[:3]
    idx = [int(cand[i]) for i in order]; scores = [float(sc[i]) for i in order]
    ms[2] = (time.time() - t) * 1000

    note = ""
    if not guard_off and conf < tau:
        if scores[0] >= RR_RESCUE:
            note = '<span class="chip">rescued by reranker</span>'
        else:
            hist.append({"q": question, "ok": False, "conf": conf, "secs": time.time() - t_all})
            yield (refuse_card(conf, tau, scores[0]),
                   pipeline_html(["stop", "done", "done", "skip"], ms),
                   sources_html(idx, scores, question, "Closest passages (not used)"),
                   history_html(hist), hist)
            return
    if guard_off:
        note = '<span class="chip">guard off</span>'

    src = sources_html(idx, scores, question)
    yield THINKING, pipeline_html(["done", "done", "done", "active"], ms), src, history_html(hist), hist
    t = time.time()
    ctx = "\n\n".join(BEST.texts[i] for i in idx)
    prompt = ("Read the context below and answer the question in one complete sentence. "
              f"Use only the context.\n\nContext:\n{ctx}\n\nQuestion: {question}\nAnswer:")
    out = gen(prompt); ms[3] = (time.time() - t) * 1000
    secs = time.time() - t_all
    hist.append({"q": question, "ok": True, "conf": conf, "secs": secs})
    yield (answer_card(out, conf, len(idx), secs, note), pipeline_html(["done"] * 4, ms), src,
           history_html(hist), hist)

def clear_all():
    return "", IDLE, pipeline_html(["pending"] * 4, [None] * 4), "", history_html([]), []

def rebuild_ui(file):
    global BEST, pages
    if file is None:
        return "Waiting for a PDF...", stats_html()
    path = file if isinstance(file, str) else file.name
    pages = load_pages(path)
    BEST = Index(chunk(pages, CHUNK_SIZE))
    return f"✅ Indexed {len(pages)} pages into {len(BEST.texts)} chunks", stats_html()

EXAMPLES = ["What is the minimum attendance required?",
            "How many activity points must regular UG students earn?",
            "When is the I grade awarded?",
            "What is the plagiarism limit for project reports?",
            "What are the canteen timings?"]

# ------------------------------------------------------------ layout
try: demo.close()
except Exception: pass

with gr.Blocks(title="DocuQA") as demo:
    hist_state = gr.State([])
    gr.HTML(HERO)
    stats = gr.HTML(stats_html())

    with gr.Tabs():
        with gr.Tab("💬 Ask"):
            with gr.Row():
                with gr.Column(scale=7):
                    q = gr.Textbox(show_label=False, lines=1,
                                   placeholder="Ask anything about the document, e.g. What happens if my attendance is below 85%?")
                    with gr.Row():
                        tau = gr.Slider(0.30, 0.85, value=0.60, step=0.01, label="Refusal threshold τ (lower = answers more)", scale=3)
                        guard_off = gr.Checkbox(label="Disable guard (always answer)", value=False, scale=2)
                    with gr.Row():
                        btn = gr.Button("Ask ✨", variant="primary", scale=3)
                        clr = gr.Button("Clear", variant="secondary", scale=1)
                    gr.Examples(EXAMPLES, inputs=q, label="Try these")
                    pipe = gr.HTML(pipeline_html(["pending"] * 4, [None] * 4))
                    ans = gr.HTML(IDLE)
                    srcs = gr.HTML("")
                with gr.Column(scale=3):
                    gr.HTML('<div class="sec" style="margin-top:0">Session history</div>')
                    hist_view = gr.HTML(history_html([]))
                    with gr.Accordion("📎 Use a different PDF", open=False):
                        f = gr.File(label="PDF", file_types=[".pdf"])
                        status = gr.Textbox(label="Status", interactive=False, value="Using syllabus.pdf")
        with gr.Tab("📊 Benchmarks"):
            gr.HTML(benchmarks_html())
            gr.HTML(threshold_svg())
        with gr.Tab("🧩 How it works"):
            gr.HTML(HOW)

    outs = [ans, pipe, srcs, hist_view, hist_state]
    btn.click(ask, [q, hist_state, tau, guard_off], outs)
    q.submit(ask, [q, hist_state, tau, guard_off], outs)
    clr.click(clear_all, None, [q, ans, pipe, srcs, hist_view, hist_state])
    f.upload(rebuild_ui, f, [status, stats])

demo.launch(share=True,
            theme=gr.themes.Base(primary_hue="violet", secondary_hue="cyan", neutral_hue="slate"),
            css=CSS)
```
